```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from IPython.display import display, Markdown

todays_date = datetime.now().strftime("%Y-%m-%d")
student_id = "jacgra1349"

# Display assignment information
display(Markdown("### Scenario Analysis - Loan Repayment"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))
```

```python
display(Markdown("### Step 2.1: Define Loan Repayment Scenarios"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define loan parameters
principal = 300000

# Define scenario parameters
scenarios = [
    {"Scenario": "Scenario 1", "Interest_Rate": 0.03, "Loan_Term": 15},
    {"Scenario": "Scenario 2", "Interest_Rate": 0.05, "Loan_Term": 30},
    {"Scenario": "Scenario 3", "Interest_Rate": 0.08, "Loan_Term": 40}
]

# Define function to calculate monthly loan repayment
def calculate_monthly_payment(P, annual_rate, years):
    r = annual_rate / 12
    n = years * 12
    M = P * r * (1 + r)**n / ((1 + r)**n - 1)
    return M

# Display scenario parameters
print("Loan Principal:", principal)
print("Loan Repayment Scenarios:")
display(pd.DataFrame(scenarios))
```

```python
display(Markdown("### Step 2.2: Perform Scenario Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Perform scenario analysis
results = []

for scenario in scenarios:
    monthly_payment = calculate_monthly_payment(
        principal,
        scenario["Interest_Rate"],
        scenario["Loan_Term"]
    )

    total_repayment = monthly_payment * scenario["Loan_Term"] * 12

    results.append({
        "Scenario": scenario["Scenario"],
        "Interest_Rate": scenario["Interest_Rate"],
        "Loan_Term": scenario["Loan_Term"],
        "Monthly_Payment": monthly_payment,
        "Total_Repayment": total_repayment
    })

# Convert results to DataFrame
results_df = pd.DataFrame(results)

# Display results
display(results_df)
```

```python
display(Markdown("### Step 3.1: Visualize Total Loan Repayment"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Plot total repayment amounts across scenarios
plt.figure(figsize=(12, 6))
sns.barplot(data=results_df, x="Scenario", y="Total_Repayment")

plt.title("Comparison of Total Loan Repayment Across Scenarios")
plt.xlabel("Loan Repayment Scenarios")
plt.ylabel("Total Repayment Amount ($)")
plt.grid(True, axis="y")
plt.show()
```

```python
display(Markdown("### Step 4.1: Apply Sensitivity Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Define sensitivity analysis parameters
interest_rate_scenarios = [0.03, 0.04, 0.05, 0.06, 0.07, 0.08]
loan_term = 30

# Perform sensitivity analysis
sensitivity_results = []

for interest_rate in interest_rate_scenarios:
    monthly_payment = calculate_monthly_payment(
        principal,
        interest_rate,
        loan_term
    )

    total_repayment = monthly_payment * loan_term * 12

    sensitivity_results.append({
        "Interest_Rate": interest_rate,
        "Monthly_Payment": monthly_payment,
        "Total_Repayment": total_repayment
    })

# Convert results to DataFrame
sensitivity_df = pd.DataFrame(sensitivity_results)

# Display sensitivity analysis results
display(sensitivity_df)
```

```python
display(Markdown("### Step 4.2: Visualize Sensitivity Analysis"))
display(Markdown(f"Date: {todays_date}"))
display(Markdown(f"Student ID: {student_id}"))

# Plot the impact of interest rate on total repayment
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=sensitivity_df,
    x="Interest_Rate",
    y="Total_Repayment",
    marker="o"
)

plt.xticks(
    interest_rate_scenarios,
    [f"{rate:.0%}" for rate in interest_rate_scenarios]
)

plt.title("Impact of Interest Rate Changes on Total Loan Repayment")
plt.xlabel("Annual Interest Rate")
plt.ylabel("Total Repayment Amount ($)")
plt.grid(True)
plt.show()
```

1.	What differences were observed between the scenarios?
The three scenarios differed based on interest rates and repayment periods. The scenario with a 3% interest rate and a 15-year repayment period had the lowest repayment amount of $372 914.09. Scenario 2 with a 5% interest rate and a 30-year repayment period had $579,767.35, and lastly, scenario 3 with an 8% interest rate and a 40-year repayment period had $ 1,001,248.83. This shows that higher interest rates and longer repayment periods lead to a higher total repayment amount.
2.	How did changes in key variables affect the outcomes in the sensitivity analysis?

When the annual interest rate was increased, monthly payments and total loan repayment increased. The interest rate was varied from 3% to 8%, while the principal amount remained at $ 300,000 and the repayment period was 30 years. At a 3% interest rate, the total loan repayment was about $455332.36. At 8%, however, the total repayment increased to $792,465.74; monthly payments also increased from $ 1,264.81 to $ 2,201.29. Thus, increasing interest rates lead to higher borrowing costs.
